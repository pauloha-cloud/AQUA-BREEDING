```
%%writefile app.py
import streamlit as st
import numpy as np
import plotly.graph_objects as go

# 1. Page Configuration
st.set_page_config(
    page_title="GxE Genotyping Simulator",
    layout="wide",
    initial_sidebar_state="expanded"
)

st.title("🐟 Simulator: Standard Error of Genetic Correlation")
st.markdown("""
Adjust the genetic parameters in the sidebar to visualize how many animals
need to be genotyped per environment to reach an acceptable Standard Error (SE ≤ 0.15).
""")

# 2. Input Interface (Sidebar)
st.sidebar.header("Experiment Parameters")

h1 = st.sidebar.slider(
    "Heritability - Environment 1 (h²)",
    min_value=0.05, max_value=0.60, value=0.30, step=0.05,
    help="Proportion of phenotypic variance explained by genetics in Environment 1."
)

h2 = st.sidebar.slider(
    "Heritability - Environment 2 (h²)",
    min_value=0.05, max_value=0.60, value=0.30, step=0.05,
    help="Proportion of phenotypic variance explained by genetics in Environment 2."
)

rg = st.sidebar.slider(
    "Expected Genetic Correlation (rg)",
    min_value=0.10, max_value=0.95, value=0.60, step=0.05,
    help="Expected correlation between the two environments (GxE)."
)

# 3. Calculation Logic
THRESHOLD = 0.15  # Acceptable Standard Error limit
K = 0.6           # Empirical adjustment constant for half-sib families

# Animal count vector from 100 to 3000
N = np.arange(100, 3050, 50)

# SE approximation formula
se = ((1 - rg**2) / np.sqrt(N)) * (1 / np.sqrt(h1 * h2)) * K

# Calculating exact algebraic intersection (where SE = 0.15)
n_intersect = (((1 - rg**2) * K) / (THRESHOLD * np.sqrt(h1 * h2)))**2

# 4. Interactive Plot with Plotly
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=N, y=se,
    mode='lines',
    name='Error Curve (SE)',
    line=dict(color='#1f77b4', width=4),
    hovertemplate='N: %{x}<br>SE: %{y:.3f}<extra></extra>'
))

fig.add_hline(
    y=THRESHOLD,
    line_dash="dash",
    line_color="red",
    annotation_text="Tolerance Limit (SE = 0.15)",
    annotation_position="top right",
    annotation_font_color="red"
)

if 100 <= n_intersect <= 3000:
    fig.add_trace(go.Scatter(
        x=[n_intersect], y=[THRESHOLD],
        mode='markers+text',
        name='Ideal Point',
        text=[f"N ≈ {int(n_intersect)}"],
        textposition="top right",
        textfont=dict(color="red", size=14),
        marker=dict(color='red', size=12, symbol='circle')
    ))

fig.update_layout(
    xaxis_title="Number of Genotyped Animals (Per Environment)",
    yaxis_title="Estimated Standard Error (SE)",
    yaxis_range=[0, 0.8],
    xaxis_range=[100, 3000],
    hovermode="x unified",
    margin=dict(l=20, r=20, t=40, b=20)
)

# 5. Main Screen Display
st.plotly_chart(fig, use_container_width=True)

st.subheader("💡 Conclusion for Design")
if n_intersect > 3000:
    st.error(f"**ALERT:** Heritability or correlation is too low. You will need approximately **{int(n_intersect)} animals per environment** to reach an error of 0.15. This might be financially unfeasible.")
elif n_intersect < 100:
    st.success("Favorable scenario. Fewer than **100 animals** already provide high precision.")
else:
    st.success(f"The safety limit (SE = 0.15) will be reached by genotyping approximately **{int(n_intersect)} animals** in each environment.")

st.info("Methodological Note: Calculations use a relationship constant assuming standard family structures in Tilapia breeding programs.")
```
